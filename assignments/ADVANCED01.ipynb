**Задача:** превратить карту интенсивности полученную с двумерного детектора в распределение интенсивности по углy рассеяния $2\Theta$, проинтегрировав по углу поворота вокруг прямого пучка ($\varphi$). Картина рассеяния должна выглядеть примерно так:

```python
from PIL import Image
import numpy as np
from matplotlib import pyplot as plt

# читаем изображение из файла
dd = Image.open("datasets/19Ti64_cut-00015.tif")
# превращаем изображение в numpy-массив
dd = np.array(dd)
# строим график (логарифмическая шкала интенсивности для лучшей видимости, по осям XY индексы пикселей)
plt.imshow(np.log(dd))
```
